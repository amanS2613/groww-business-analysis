```python
%sql
SELECT
    COALESCE(acquisition_channel, 'Unknown') AS acquisition_channel,

    COUNT(*) AS total_users,

    SUM(high_value_90d) AS high_value_users,

    ROUND(
        100.0 * SUM(high_value_90d) / COUNT(*), 2
    ) AS high_value_rate_pct,

    ROUND(SUM(observable_revenue_90d), 2) AS total_revenue,

    ROUND(
        AVG(observable_revenue_90d), 2
    ) AS revenue_per_user,

    ROUND(
        AVG(
            CASE
                WHEN high_value_90d = 1
                THEN observable_revenue_90d
            END
        ), 2
    ) AS revenue_per_high_value_user

FROM default.groww_modeling_dataset

GROUP BY COALESCE(acquisition_channel, 'Unknown')

ORDER BY revenue_per_user DESC;
```

```python
%sql
SELECT
    COALESCE(acquisition_channel, 'Unknown') AS acquisition_channel,
    COALESCE(acquisition_subchannel, 'Unknown') AS acquisition_subchannel,

    COUNT(*) AS total_users,
    SUM(high_value_90d) AS high_value_users,

    ROUND(
        100.0 * SUM(high_value_90d) / COUNT(*), 2
    ) AS high_value_rate_pct,

    ROUND(
        AVG(observable_revenue_90d), 2
    ) AS revenue_per_user,

    ROUND(
        SUM(observable_revenue_90d), 2
    ) AS total_revenue

FROM default.groww_modeling_dataset

GROUP BY
    COALESCE(acquisition_channel, 'Unknown'),
    COALESCE(acquisition_subchannel, 'Unknown')

HAVING COUNT(*) >= 30

ORDER BY revenue_per_user DESC;
```
